# Tabular data

```python
import seaborn as sns

df = sns.load_dataset("penguins")

df.describe()
```

```python
sns.scatterplot(x=df['bill_length_mm'], y=df['bill_depth_mm'], hue=df['species'])
```

```python
sns.scatterplot(data=df, x='bill_length_mm', y='bill_depth_mm', hue='species', size='body_mass_g')
```

```python
sns.histplot(x=df['body_mass_g'])
```

```python
sns.histplot(x=df['body_mass_g'], bins=30)
```

```python
sns.histplot(x=df['body_mass_g'], bins=30, hue=df['species'], multiple='stack')
```

```python
sns.histplot(x=df['body_mass_g'], bins=30, hue=df['species'], multiple='dodge')
```

```python
sns.histplot(x=df['body_mass_g'], bins=30, hue=df['species'], multiple='layer')
```

```python
sns.histplot(x=df['body_mass_g'], bins=30, hue=df['species'], multiple='fill')
```

```python
sns.boxplot(y=df['body_mass_g'], hue=df['species'])
```

```python
sns.kdeplot(x=df['body_mass_g'], hue=df['species'])
```

```python
sns.displot(x=df['body_mass_g'], hue=df['species'], kind='kde', bw_adjust=0.5)
```

```python
sns.violinplot(y=df['body_mass_g'], hue=df['species'])
```

```python
sns.pairplot(df, hue="species")
```
