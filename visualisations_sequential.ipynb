```python
import seaborn as sns

sns.get_dataset_names()
```

```python
df = sns.load_dataset('fmri')

df
```

```python
df['subject'].unique()
```

```python
df_by_subject = df.groupby(by='subject')
df_aux = df_by_subject.get_group('s0')
ax = sns.scatterplot(data=df_aux, x='timepoint', y='signal',
                hue=df_aux[['region', 'event']].apply(tuple, axis=1))
ax.get_legend().set_title('region and event')
```

```python
ax = sns.lineplot(data=df_aux, x='timepoint', y='signal',
             hue=df_aux[['region', 'event']].apply(tuple, axis=1),
             marker='o')
ax.get_legend().set_title('region and event')
```

```python
ax = sns.lineplot(data=df_aux, x='timepoint', y='signal',
             hue=df_aux[['region', 'event']].apply(tuple, axis=1))
ax.get_legend().set_title('region and event')
```

```python
df_by_event_region = df.groupby(by=['event', 'region'])
df_aux = df_by_event_region.get_group(('stim', 'parietal'))
ax = sns.scatterplot(data=df_aux, x='timepoint', y='signal')
ax.set_title('event = stim, region = parietal')
```

```python
df_by_event_region = df.groupby(by=['event', 'region'])
df_aux = df_by_event_region.get_group(('stim', 'parietal'))
ax = sns.lineplot(data=df_aux, x='timepoint', y='signal')
ax.set_title('event = stim, region = parietal')
```
